# Naïve Bayes Models

```python
import pandas as pd
df = pd.read_csv("music_data.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
# Remove duplicates
df = df.drop_duplicates()
df.size
```

```python
df.isnull().sum()
```

```python
# Fill missing numeric values with median

df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder

categorical_cols = df.select_dtypes(include=['object']).columns

# Create a LabelEncoder instance
le = LabelEncoder()

# Convert each object column to numeric using Label Encoding
for column in categorical_cols:
    df[column] = le.fit_transform(df[column])
```

```python
print("Data after cleaning:")
df.head()
```

# ML Modelling

```python
#Input

x = df.drop(['track_popularity','track_artist',	'track_album_id',	'track_album_name',	'playlist_name','playlist_id','playlist_subgenre','loudness'],axis=1)
x
```

```python
# Output

# 1 -> Popular
# 0 -> Not Popular

y = (df['track_popularity'] >= 50).astype(int)
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print('xtrain',x_train.shape)
print('xtest',x_test.shape)
```

```python
print('ytrain',y_train.shape)
print('ytest',y_test.shape)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.naive_bayes import MultinomialNB, GaussianNB, BernoulliNB

mnb = MultinomialNB()
mnb.fit(x_train, y_train)  # Use scaled training data
y_pred_mnb = mnb.predict(x_test)  # Use scaled testing data

print("Multinomial NB:\n", classification_report(y_test, y_pred_mnb))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_mnb), annot=True, cmap='Blues', fmt='d')
plt.title('Multinomial NB Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Gaussian Naive Bayes

gnb = GaussianNB()
gnb.fit(x_train, y_train)
y_pred_gnb = gnb.predict(x_test)

print("Gaussian NB:\n", classification_report(y_test, y_pred_gnb))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_gnb), annot=True, cmap='Blues', fmt='d')
plt.title('Gaussian NB Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Bernoulli Naive Bayes

bnb = BernoulliNB()
bnb.fit(x_train, y_train)
y_pred_bnb = bnb.predict(x_test)

print("Bernoulli NB:\n", classification_report(y_test, y_pred_bnb))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_bnb), annot=True, cmap='Blues', fmt='d')
plt.title('Bernoulli NB Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import KBinsDiscretizer

# Categorical Naive Bayes

kbins = KBinsDiscretizer(n_bins=5, encode='ordinal', strategy='uniform')
x_train_binned = kbins.fit_transform(x_train)
x_test_binned = kbins.transform(x_test)

# Fit Categorical Naive Bayes
cnb = CategoricalNB()
cnb.fit(x_train_binned, y_train)
y_pred_cnb = cnb.predict(x_test_binned)

# Evaluation for Categorical NB
print("Categorical Naive Bayes:\n", classification_report(y_test, y_pred_cnb))
```

```python
# Plot confusion matrix
sns.heatmap(confusion_matrix(y_test, y_pred_cnb), annot=True, cmap='Blues', fmt='d')
plt.title('Categorical Naive Bayes Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
