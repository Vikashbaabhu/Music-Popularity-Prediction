# Decision Tree Models

```python
import pandas as pd
df = pd.read_csv("music_data.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
# Remove duplicates
df = df.drop_duplicates()
df.size
```

```python
df.isnull().sum()
```

```python
# Fill missing numeric values with median

df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder

categorical_cols = df.select_dtypes(include=['object']).columns

# Create a LabelEncoder instance
le = LabelEncoder()

# Convert each object column to numeric using Label Encoding
for column in categorical_cols:
    df[column] = le.fit_transform(df[column])
```

```python
print("Data after cleaning:")
df.head()
```

# ML Modelling

```python
#Input

x = df.drop(['track_popularity','track_artist',	'track_album_id',	'track_album_name',	'playlist_name','playlist_id','playlist_subgenre','loudness'],axis=1)
x
```

```python
# Output

# 1 -> Popular
# 0 -> Not Popular

y = (df['track_popularity'] >= 50).astype(int)
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
print('xtrain',x_train.shape)
print('xtest',x_test.shape)
```

```python
print('ytrain',y_train.shape)
print('ytest',y_test.shape)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Decision Tree (Entropy)
dt_entropy = DecisionTreeClassifier(criterion='entropy', random_state=42)
dt_entropy.fit(x_train, y_train)

y_pred_dt_entropy = dt_entropy.predict(x_test)

print("Decision Tree (Entropy):\n", classification_report(y_test, y_pred_dt_entropy))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_dt_entropy), annot=True, cmap='Blues', fmt='d')
plt.title('Decision Tree (Entropy) Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Plot Entropy-based tree

plt.figure(figsize=(20, 10))
plot_tree(dt_entropy, filled=True, feature_names=x.columns,class_names=['Not Popular', 'Popular'],impurity=True,proportion=True,rounded=True)
plt.title('Decision Tree (Entropy)')
plt.show()
```

```python
# Decision Tree (Gini)
dt_gini = DecisionTreeClassifier(criterion='gini', random_state=42)
dt_gini.fit(x_train, y_train)

y_pred_dt_gini = dt_gini.predict(x_test)

print("Decision Tree (Gini):\n", classification_report(y_test, y_pred_dt_gini))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_dt_gini), annot=True, cmap='Blues', fmt='d')
plt.title('Decision Tree (Gini) Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Plot Gini-based tree
plt.figure(figsize=(20, 10))
plot_tree(dt_gini, filled=True, feature_names=x.columns, class_names=['Not Popular', 'Popular'])
plt.title('Decision Tree (Gini)')
plt.show()
```

```python
# Decision Tree (log_loss)
dt_loss = DecisionTreeClassifier(criterion='log_loss', random_state=42)
dt_loss.fit(x_train, y_train)

y_pred_dt_loss = dt_loss.predict(x_test)

print("Decision Tree (log_loss):\n", classification_report(y_test, y_pred_dt_loss))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_dt_loss), annot=True, cmap='Blues', fmt='d')
plt.title('Decision Tree (log_loss) Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Plot log_loss-based tree
plt.figure(figsize=(20, 10))
plot_tree(dt_loss, filled=True, feature_names=x.columns, class_names=['Not Popular', 'Popular'])
plt.title('Decision Tree (log_loss)')
plt.show()
```
